```python
tabla = int(input("Dame la tabla de multiplicar que quieres ver "))
v1 = int(input("Dame el primer valor "))
v2 = int(input("Dame el segundo valor "))
while v2 <= v1:
  print("El primer valor debe ser menor que el segundo")
  v2 = int(input("Dame el segundo valor "))
v3 = int(input("Dame el tercer valor "))
for i in range(v1,v2,v3):
  print(tabla,"x",i,"=",tabla*i)
```
