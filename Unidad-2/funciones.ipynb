**Regresar un resultado - la instruccion Return**

```python
def calcular_area(base, altura):
  area = base * altura
  return area
```

#**Alcance de las variables**
- El alcance de una variable se refiere a la parte del codigo donde se conoce la variable
- Una variable que se define dentro de una funcion no existe en otras funciones o fuera de la funcion
- Para extender el alcance de una variable y que pueda existir fuera de una funcion se debe definir con la palabra reservada **global**

```python
def mifuncion():
  variable_x = 123
  print(variable_x)
mifuncion()
print(variable_x)
```
