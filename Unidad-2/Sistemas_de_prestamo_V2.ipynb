```python
#SISTEMA DE PRÉSTAMOS
# Versión 1 Programación basica y uso de listas
# Lista de alumnos activos en el sistema
import datetime
from datetime import datetime, timedelta
alumnos = [
           ["26040200", "Hernesto", "Ingenieria en Inteligencia Artificial"],
           ["26040201", "Javier", "Ingenieria en Inteligencia Artificial"],
           ["26040202", "Dana", "Ingenieria en Informatica"],
           ["26040203", "Simon", "Ingenieria en Informatica"],
           ["26040204", "Abiel", "Ingenieria en Inteligencia Artificial"]
           ]
# Cada valor en la lista de computadoras tiene su lista correspondiente con el aula donde esta ubicada, el software instalado y el estado de esa computadora computadoras = ["PC-LC1-01", "PC-LC1-02", "PC-LC1-03", "PC-LC2-01", "PC-LC2-02", "PC-LC2-03", "PC-LC3-01", "PC-LC3-02", "PC-LC3-03", "PC-LC3-04"]
computadoras = [
               ["PC-LC1-01","LC1","Python, VS Code","Disponible",],
               ["PC-LC1-02","LC1","Python, Java, VS Code","Prestada",],
               ["PC-LC1-03","LC1","AutoCAD, MATLAB","Reservada",],
               ["PC-LC2-01","LC2","Python, VS Code","Disponible",],
               ["PC-LC2-02","LC2","Python, Java, VS Code","Prestada",],
               ["PC-LC2-03","LC2","AutoCAD, MATLAB","Reservada",],
               ["PC-LC3-01","LC3","Python, VS Code","Disponible",],
               ["PC-LC3-02","LC3","Python, Java, VS Code","Prestada",],
               ["PC-LC3-03","LC3","AutoCAD, MATLAB","Reservada",],
               ["PC-LC3-04", "LC3", "Python, VS Code","Mantenimiento"]
                ]
prestamos = []

fechas = []

programas_necesidades = []

# FUNCIONES
#1. Consultar computadoras
# si el usuario selecciono la opcion 1 se hace un ciclo (for) que recorre
#toda la lista de computadoras e imprime los valores de las demás listas
def consultar_computadoras():
    print("\n--- COMPUTADORAS ---")
    for i in range(len(computadoras)):
        print("|Aula:", computadoras[i][1], "|ID:", computadoras [i][0],
                "|Software:", computadoras[i][2], "|Estado:", computadoras [i][3])

def buscar_sw():
  programa = input( "\nIngrese el software que desea buscar: ")
  print("\n--- RESULTADOS ---")
  encontrado = False
  for i in range(len(computadoras)):
      # la funcion (lower) convierte a minuscula el texto, de esta forma
      # nos aseguramos que el texto a buscar y la informacion de las listas
      # este en minuscula y no tengamos problema de que no encunentra un programa
      # porque este escrito de forma diferente usando mayusculas y minusculas
      if programa.lower() in computadoras[i][2].lower():
          print("|ID:", computadoras[i][0], "|Aula:", computadoras[i][1],
                "|Software:", computadoras[i][2], "|Estado:", computadoras[i][3])

          encontrado = True

  if encontrado == False:
      print("No se encontraron computadoras con ese software.")

def solicitar_prestamo():
    print("\n=== SOLICITAR PRESTAMO ===")
    num_control = input("Numero de control del alumno: ")

    # --- VALIDACIÓN DEL ALUMNO ---
    alumno_valido = False
    for a in alumnos:
        if a[0] == num_control:
            alumno_valido = True
            break

    if not alumno_valido:
        print("El alumno no existe en la lista")
        return          # ← corta aquí, no ejecuta lo demás

    # --- A PARTIR DE AQUÍ, EL ALUMNO YA ES VÁLIDO ---
    programa = input("Que programa necesita: ").lower()
    software_encontrado = False

    for pc in computadoras:
        if programa in pc[2].lower():
            software_encontrado = True
            if pc[3] == "Disponible":
                pc[3] = "Prestada"
                print(f"Se asignó la computadora {pc[0]} del aula {pc[1]}")

                prestamos.append([num_control, pc[0], pc[1], pc[2]])

                fecha_prestamo = datetime.now()
                fecha_limite   = fecha_prestamo + timedelta(days=3)
                fechas.append(fecha_limite)

                print(f"Computadora prestada el: {fecha_prestamo.strftime('%d/%m/%Y %H:%M')}")
                print(f"Fecha límite: {fecha_limite.strftime('%d/%m/%Y %H:%M')}")
                return
            else:
                print(f"Se encontró el software pero la computadora está {pc[3]}")

    if not software_encontrado:
        print(f"el software '{programa}' no esta instalado en ninguna computadora.")
        confirmar = input("¿Registrarlo como necesidad? s/n ")
        if confirmar == "s":
          confirmar_2 = any(p == programa and nc == num_control
                            for p, nc, _ in programas_necesidades)
          if not confirmar_2:
            programas_necesidades.append((programa, num_control, datetime.now()))
            print(f"Se registró el software '{programa}' como una necesidad.")
          else:
            print(f"El software '{programa}' ya está registrado como una necesidad.")
        else:
          print("No se registró el software como una necesidad (error de escritura)")

def devolver_computadora():
  print("\n ===DEVOLVER COMPUTADORA====")
  id_pc = input("Dame el ID de la computadora: ")
  existe = False
  for i in range( len(computadoras) ):
    if computadoras[i][0] == id_pc:
      existe = True
      if computadoras[i][3] == "Prestada" or computadoras[i][3]=="Reservada":
        computadoras[i][3] = "Disponible"
        print("La computadora", id_pc, "ya esta disponible")
        break
      else:
        print("La computadora", id_pc, "no esta prestada/reservada")
  if existe == False:
    print("No existe la computadora", id_pc)

def agregar_computadora():
  print("\n--- AGREGAR COMPUTADORA ---")
  id_computadora = input("ID de la computadora: ")
  aula = input("Aula: ")
  programas = input("Software instalado: ")
  nueva_pc = [id_computadora, aula, programas, "Disponible"]
  computadoras.append(nueva_pc)
  print("\nComputadora agregada correctamente.")

def historial_prestamos():
  print("\n--- HISTORIAL DE PRESTAMOS ---")
  for i in range(len(prestamos)):
    print("|Alumno:", prestamos[i][0], "|Computadora:" ,prestamos[i][1], "|", prestamos[i][2], "|", prestamos[i][3], "|", fechas[i])

from collections import Counter, defaultdict

def necesidades_software():
    print("\n--- PROGRAMAS NECESARIOS ---")

    if not programas_necesidades:
        print("No hay programas pendientes por instalar.")
        return

    # Agrupamos extrayendo los 3 datos
    agrupado = defaultdict(list)
    for prog, num_control, fecha in programas_necesidades:
        agrupado[prog].append((num_control, fecha))

    print(f"\nTotal de solicitudes registradas: {len(programas_necesidades)}")
    print(f"Programas distintos solicitados:   {len(agrupado)}\n")

    # Ordenar por cantidad de solicitudes (mayor a menor)
    ordenados = sorted(agrupado.items(), key=lambda x: len(x[1]), reverse=True)

    for prog, solicitudes in ordenados:
        cantidad = len(solicitudes)
        fechas_prog = [f for _, f in solicitudes]
        primera = min(fechas_prog)
        ultima  = max(fechas_prog)

        print(f"▶ {prog.upper()}  ({cantidad} solicitud{'es' if cantidad > 1 else ''})")
        print(f"   Primera solicitud: {primera.strftime('%d/%m/%Y %H:%M')}")
        print(f"   Última  solicitud: {ultima.strftime('%d/%m/%Y %H:%M')}")

        # Detalle: quién pidió y cuándo
        for nc, fecha in sorted(solicitudes, key=lambda x: x[1]):
            print(f"   - Alumno {nc} lo pidió el {fecha.strftime('%d/%m/%Y %H:%M')}")
        print()
# def estadisticas():

def estadisticas():
  print("\n--- ESTADISTICAS ---")
  conteo_carreras = []
  conteo_programas = []
  for i in range(len(prestamos)):
    num_control = prestamos[i][0]
    carrera = ""
    # Buscar la carrera en la lista de alumnos
    for j in range(len(alumnos)):
      if alumnos[j][0] == num_control:
        carrera = alumnos[j][2]
    # Contar la carrera
    encontrado_c = False
    for k in range(len(conteo_carreras)):
      if conteo_carreras[k][0] == carrera:
        conteo_carreras[k][1] += 1
        encontrado = True
        break
    if not encontrado_c:
      conteo_carreras.append([carrera, 1])

    encontrado_p = False
    for k in range(len(conteo_programas)):
      if conteo_programas[k][0] == carrera:
        conteo_programas[k][1] += 1
        encontrado = True
        break
    if not encontrado_p :
      conteo_programas.append([carrera, 1])

# --- Mostrar resultados ---

  print("================================")
  print("Estadisticas por carrera: ")
  for i in range(len(conteo_carreras)):
    print(f"{conteo_carreras[i][0]}: {conteo_carreras[i][1]}", "veces")

  print("Estadisticas por programa: ")
  for i in range(len([])):
    print(f"{[][i][0]}: {[][i][1]}", "veces")
# MENÚ PRINCIPAL
# Usamos un ciclo While que se repetira indefinidamente hasta que el usuario
# teclee la opcion 7 (salir) o cualquier numero superior a 7
opcion = 0
while opcion <= 9:
    print("\n================================")
    print("      SISTEMA DE PRÉSTAMOS")
    print("================================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar préstamo")
    print("4. Devolver computadora")
    print("5. Crear reservación")
    print("6. Agregar computadora")
    print("7. Historial de prestamos")
    print("8. Necesidades de software")
    print("9. Estadisticas")
    print("10. Salir")
    print("================================")
    opcion = int(input("Seleccione una opción: "))

    # 1. Consultar computadoras
    # si el usuario selecciono la opcion 1 se hace un ciclo (for) que recorre
    # toda la lista de computadoras e imprime los valores de las demás listas
    if opcion == 1:
      consultar_computadoras()

    # 2. Buscar por software
    elif opcion == 2:
      buscar_sw()

    # 3. Solicitar préstamo
    elif opcion == 3:
      solicitar_prestamo()

    #4. Devolver computadora
    elif opcion==4:
      devolver_computadora()

    # 5. Crear reservación
    elif opcion == 5:
      print("\n--- CREAR RESERVACIÓN ---")

    # 6. Agregar computadora
    elif opcion == 6:
      agregar_computadora()

    elif opcion == 7:
      historial_prestamos()

    elif opcion == 8:
      necesidades_software()

    elif opcion == 9:
      estadisticas()

    # 10. Salir
    elif opcion == 10:
        print("\nGracias por utilizar el sistema.")
    else:
        print("\nOpción no válida.")
```
